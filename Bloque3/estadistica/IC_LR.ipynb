# Longitud de Replicas

```python
# Formulas

def normal_Conocido (sigma, z, e):
    n = ((z * sigma) / e) ** 2

    # redondear para arriba
    n = np.ceil(n)
    return n


def normal_Desconocido (n_prima, s, t, e):
    n = ((s / e ) * t ) ** 2

    # redondear para arriba
    n = np.ceil(n)
    return n

def desconocida (alpha, s, e):
    
    n = (1 / alpha ) * (s / e) ** 2
    # redondear para arriba
    n = np.ceil(n)
    return n
```

```python
# LR - 1

sigma = 8 

e = 1.5

z = 1.96

n = (((sigma * z) /2) / e) ** 2

n2 = normal_Conocido (sigma, z, e)
print("Cantidad de Replicas con funcion: ", n2)
```
