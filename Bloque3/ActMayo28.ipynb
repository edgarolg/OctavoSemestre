```python
import numpy as np
import random


print("Simulación Escenario A, Problema 1")

def tiempo_cajero_1():
  t = np.random.normal(3.5, 0.8)
  while t <= 0: 
    t = np.random.normal(3.5, 0.8)
  return t

def tiempo_cajero_2():
  return np.random.uniform(4, 7)

def generacion_clientes():
  return np.random.exponential(4)

def simulacion_un_turno():
  tiempo_turno = 240
  tiempo_llegada_actual = 0
  fin_servicio_c1 = 0
  fin_servicio_c2 = 0
  
  tiempos_espera = []
  tiempos_sistema = []
  tiempo_trabajado_c1 = 0
  tiempo_trabajado_c2 = 0
  clientes_c1 = 0
  clientes_c2 = 0
  
  while True:
    tiempo_llegada_actual += generacion_clientes()
    if tiempo_llegada_actual > tiempo_turno:
        break
          
    if fin_servicio_c1 <= tiempo_llegada_actual and fin_servicio_c2 <= tiempo_llegada_actual:
      cajero_asignado = 1
    elif fin_servicio_c1 <= tiempo_llegada_actual:
      cajero_asignado = 1
    elif fin_servicio_c2 <= tiempo_llegada_actual:
      cajero_asignado = 2
    else:
      if fin_servicio_c1 <= fin_servicio_c2:
        cajero_asignado = 1
      else:
        cajero_asignado = 2
              
    if cajero_asignado == 1:
      inicio_servicio = max(tiempo_llegada_actual, fin_servicio_c1)
      tiempo_servicio = tiempo_cajero_1()
      fin_servicio_c1 = inicio_servicio + tiempo_servicio
          
      tiempo_trabajado_c1 += tiempo_servicio
      clientes_c1 += 1
    else:
      inicio_servicio = max(tiempo_llegada_actual, fin_servicio_c2)
      tiempo_servicio = tiempo_cajero_2()
      fin_servicio_c2 = inicio_servicio + tiempo_servicio
          
      tiempo_trabajado_c2 += tiempo_servicio
      clientes_c2 += 1
          
    tiempo_espera = inicio_servicio - tiempo_llegada_actual
    tiempo_sistema = tiempo_espera + tiempo_servicio
      
    tiempos_espera.append(tiempo_espera)
    tiempos_sistema.append(tiempo_sistema)
      
  promedio_espera = np.mean(tiempos_espera) if tiempos_espera else 0
  promedio_sistema = np.mean(tiempos_sistema) if tiempos_sistema else 0

  utilizacion_c1 = (tiempo_trabajado_c1 / tiempo_turno) * 100
  utilizacion_c2 = (tiempo_trabajado_c2 / tiempo_turno) * 100

  return {
    "promedio_espera": promedio_espera,
    "promedio_sistema": promedio_sistema,
    "utilizacion_c1": utilizacion_c1,
    "utilizacion_c2": utilizacion_c2,
    "clientes_c1": clientes_c1,
    "clientes_c2": clientes_c2
  }

def simular_multiples_turnos(num_turnos):
  resultados = {
    "promedio_espera": 0, "promedio_sistema": 0,
    "utilizacion_c1": 0, "utilizacion_c2": 0,
    "clientes_c1": 0, "clientes_c2": 0
  }
    
  for _ in range(num_turnos):
    res_turno = simulacion_un_turno()
    for clave in resultados:
      resultados[clave] += res_turno[clave]
            
  for clave in resultados:
    resultados[clave] /= num_turnos
        
  print(f"\n-RESULTADOS PARA {num_turnos} TURNOS-")
  print(f"1. Tiempo Promedio de Espera: {resultados['promedio_espera']:.2f} minutos")
  print(f"2. Tiempo Promedio en el Sistema: {resultados['promedio_sistema']:.2f} minutos")
  print(f"3. Utilización Cajero 1: {resultados['utilizacion_c1']:.2f}%")
  print(f"3. Utilización Cajero 2: {resultados['utilizacion_c2']:.2f}%")
  print(f"4. Desempeño: Cajero 1 atendió a {resultados['clientes_c1']:.1f} clientes en promedio.")
  print(f"4. Desempeño: Cajero 2 atendió a {resultados['clientes_c2']:.1f} clientes en promedio.")

turnos_simular = [5, 10, 30]

for n in turnos_simular:
  simular_multiples_turnos(n)
```

```python
def tiempo_kiosco():
	return np.random.uniform(2, 4)

def tiempo_entrega():
	t = np.random.normal(3, 0.5)
	while t <= 0:
		t = np.random.normal(3, 0.5)
	return t

def tiempo_tradicional():
	return np.random.uniform(5, 8)

def generacion_clientes():
	return np.random.exponential(5)

def simulacion_un_turno_b():
	tiempo_turno = 300
	tiempo_llegada_actual = 0
	
	fin_servicio_kiosco = 0
	fin_servicio_entrega = 0
	fin_servicio_tradicional = 0
	
	tiempo_trabajado_kiosco = 0
	tiempo_trabajado_entrega = 0
	tiempo_trabajado_tradicional = 0
	
	esperas_kiosco = []
	esperas_tradicional = []
	tiempos_sistema = []
	esperas_mas_de_8 = 0
	
	while True:
		tiempo_llegada_actual += generacion_clientes()
		if tiempo_llegada_actual > tiempo_turno:
			break
			
		if random.random() < 0.60:
			inicio_kiosco = max(tiempo_llegada_actual, fin_servicio_kiosco)
			t_kiosco = tiempo_kiosco()
			fin_servicio_kiosco = inicio_kiosco + t_kiosco
			tiempo_trabajado_kiosco += t_kiosco
			espera_kiosco = inicio_kiosco - tiempo_llegada_actual
			
			llegada_entrega = fin_servicio_kiosco
			inicio_entrega = max(llegada_entrega, fin_servicio_entrega)
			t_entrega = tiempo_entrega()
			fin_servicio_entrega = inicio_entrega + t_entrega
			tiempo_trabajado_entrega += t_entrega
			espera_entrega = inicio_entrega - llegada_entrega
			
			espera_total = espera_kiosco + espera_entrega
			tiempo_total_sistema = espera_total + t_kiosco + t_entrega
			
			esperas_kiosco.append(espera_total)
			tiempos_sistema.append(tiempo_total_sistema)
			if espera_total > 8:
				esperas_mas_de_8 += 1
				
		else:
			inicio_tradicional = max(tiempo_llegada_actual, fin_servicio_tradicional)
			t_tradicional = tiempo_tradicional()
			fin_servicio_tradicional = inicio_tradicional + t_tradicional
			tiempo_trabajado_tradicional += t_tradicional

			espera = inicio_tradicional - tiempo_llegada_actual
			tiempo_total_sistema = espera + t_tradicional
			
			esperas_tradicional.append(espera)
			tiempos_sistema.append(tiempo_total_sistema)
			if espera > 8:
				esperas_mas_de_8 += 1
				
	total_clientes = len(tiempos_sistema)
	prom_espera_kiosco = np.mean(esperas_kiosco) if esperas_kiosco else 0
	prom_espera_trad = np.mean(esperas_tradicional) if esperas_tradicional else 0
	prom_sistema = np.mean(tiempos_sistema) if tiempos_sistema else 0
	pct_mas_8 = (esperas_mas_de_8 / total_clientes * 100) if total_clientes > 0 else 0
	
	util_kiosco = (tiempo_trabajado_kiosco / tiempo_turno) * 100
	util_entrega = (tiempo_trabajado_entrega / tiempo_turno) * 100
	util_tradicional = (tiempo_trabajado_tradicional / tiempo_turno) * 100
	
	return {
		"prom_espera_kiosco": prom_espera_kiosco,
		"prom_espera_trad": prom_espera_trad,
		"prom_sistema": prom_sistema,
		"util_kiosco": util_kiosco,
		"util_entrega": util_entrega,
		"util_tradicional": util_tradicional,
		"pct_mas_8": pct_mas_8
	}

def simular_multiples_turnos_b(num_turnos):
	resultados = {
		"prom_espera_kiosco": 0, "prom_espera_trad": 0, "prom_sistema": 0,
		"util_kiosco": 0, "util_entrega": 0, "util_tradicional": 0,
		"pct_mas_8": 0
	}
	
	for _ in range(num_turnos):
		res = simulacion_un_turno_b()
		for clave in resultados:
			resultados[clave] += res[clave]
			
	for clave in resultados:
		resultados[clave] /= num_turnos
		
	print(f"\n--- ESCENARIO B: RESULTADOS PARA {num_turnos} TURNOS ---")
	print(f"1. Promedio Espera (Kiosco): {resultados['prom_espera_kiosco']:.2f} min")
	print(f"1. Promedio Espera (Tradicional): {resultados['prom_espera_trad']:.2f} min")
	print(f"2. Tiempo Promedio Sistema (Global): {resultados['prom_sistema']:.2f} min")
	print(f"3. Utilización Kiosco: {resultados['util_kiosco']:.2f}%")
	print(f"3. Utilización Empleado Entrega: {resultados['util_entrega']:.2f}%")
	print(f"3. Utilización Empleado Tradicional: {resultados['util_tradicional']:.2f}%")
	print(f"4. Clientes que esperan > 8 minutos: {resultados['pct_mas_8']:.2f}%")

# Ejecución solicitada
turnos_a_simular = [5, 10, 30]

for n in turnos_a_simular:
	simular_multiples_turnos_b(n)
```

```python
def tiempo_caja_normal():
	return np.random.uniform(5, 9)

def tiempo_caja_preferente():
	t = np.random.normal(4, 0.8)
	while t <= 0:
		t = np.random.normal(4, 0.8)
	return t

def generacion_clientes():
	return np.random.exponential(6)

def simulacion_turno_b():
	tiempo_turno = 300
	tiempo_llegada_actual = 0
	
	fin_servicio_normal = 0
	fin_servicio_pref = 0
	
	tiempo_trabajado_normal = 0
	tiempo_trabajado_pref = 0
	
	esperas_normal = []
	esperas_pref = []
	tiempos_sistema_totales = []
	
	while True:
		tiempo_llegada_actual += generacion_clientes()
		if tiempo_llegada_actual > tiempo_turno:
			break
			
		if random.random() < 0.75:
			inicio_servicio = max(tiempo_llegada_actual, fin_servicio_normal)
			t_servicio = tiempo_caja_normal()
			fin_servicio_normal = inicio_servicio + t_servicio
			tiempo_trabajado_normal += t_servicio
			
			espera = inicio_servicio - tiempo_llegada_actual
			esperas_normal.append(espera)
			tiempos_sistema_totales.append(espera + t_servicio)
		else:
			inicio_servicio = max(tiempo_llegada_actual, fin_servicio_pref)
			t_servicio = tiempo_caja_preferente()
			fin_servicio_pref = inicio_servicio + t_servicio
			tiempo_trabajado_pref += t_servicio
			
			espera = inicio_servicio - tiempo_llegada_actual
			esperas_pref.append(espera)
			tiempos_sistema_totales.append(espera + t_servicio)
			
	total_clientes = len(tiempos_sistema_totales)
	prom_espera_normal = np.mean(esperas_normal) if esperas_normal else 0
	prom_espera_pref = np.mean(esperas_pref) if esperas_pref else 0
	promedio_sistema = np.mean(tiempos_sistema_totales) if total_clientes > 0 else 0
	
	utilizacion_normal = (tiempo_trabajado_normal / tiempo_turno) * 100
	utilizacion_pref = (tiempo_trabajado_pref / tiempo_turno) * 100
	
	return {
		"prom_espera_normal": prom_espera_normal,
		"prom_espera_pref": prom_espera_pref,
		"promedio_sistema": promedio_sistema,
		"utilizacion_normal": utilizacion_normal,
		"utilizacion_pref": utilizacion_pref,
		"total_clientes": total_clientes
	}

def simular_multiples_turnos_b(num_turnos):
	resultados = {
		"prom_espera_normal": 0, "prom_espera_pref": 0, "promedio_sistema": 0,
		"utilizacion_normal": 0, "utilizacion_pref": 0, "total_clientes": 0
	}
	
	for _ in range(num_turnos):
		res_turno = simulacion_turno_b()
		for clave in resultados:
			resultados[clave] += res_turno[clave]
			
	for clave in resultados:
		resultados[clave] /= num_turnos
		
	print(f"\n--- ESCENARIO B (BANCO - CAJAS SEPARADAS): RESULTADOS PARA {num_turnos} TURNOS ---")
	print(f"1. Promedio Espera (Clientes Regulares): {resultados['prom_espera_normal']:.2f} min")
	print(f"1. Promedio Espera (Clientes Preferentes): {resultados['prom_espera_pref']:.2f} min")
	print(f"2. Tiempo Promedio Total en el Sistema: {resultados['promedio_sistema']:.2f} min")
	print(f"3. Utilización Caja Normal: {resultados['utilizacion_normal']:.2f}%")
	print(f"3. Utilización Caja Preferente: {resultados['utilizacion_pref']:.2f}%")
	print(f"-  Total de clientes atendidos por turno: {resultados['total_clientes']:.1f}")

turnos_a_simular = [5, 10, 30]

for n in turnos_a_simular:
	simular_multiples_turnos_b(n)
```
